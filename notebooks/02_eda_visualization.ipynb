### Show me real examples 🖼️
Nothing builds intuition like looking at actual patches. The helper below lays out a few
(pre, post) pairs per damage level. 👉 *TODO(Role B):* run it and eyeball the results.

```python
from PIL import Image

def show_examples(manifest, n_per_class=3):
    """Grid of (pre, post) pairs, one row per damage class."""
    classes = sorted(manifest['label'].unique())
    fig, axes = plt.subplots(len(classes), n_per_class*2, figsize=(3*n_per_class, 2*len(classes)))
    for r, c in enumerate(classes):
        rows = manifest[manifest.label==c].sample(min(n_per_class, (manifest.label==c).sum()))
        for j, (_, row) in enumerate(rows.iterrows()):
            axes[r, 2*j].imshow(Image.open(row.pre_path));   axes[r, 2*j].set_title('pre')
            axes[r, 2*j+1].imshow(Image.open(row.post_path)); axes[r, 2*j+1].set_title(config.IDX_TO_NAME[c])
    for ax in axes.ravel(): ax.axis('off')
    plt.tight_layout(); plt.show()

# show_examples(manifest)
```
