# 00 · Project Overview & Task Division

**Mapping Natural-Disaster Damage from Satellite Imagery**  
Team: Bhuvanesh Dinesh Wadhwani · Artur Zavistovskyi · Anastasiia Khitrova · Yitian Zhou

Core idea = **change detection**: compare *pre* and *post* disaster imagery and classify
building damage on a 4-level scale (no damage / minor / major / destroyed).

The headline research question: **does a model trained on one disaster type transfer to another?**
We therefore evaluate with a **held-out disaster type**, not a random split.

## Grading (30 pts total)

| Part | Points | What graders look for |
|------|--------|------------------------|
| 1 · Data & Preliminary Analysis | 10 | co-registration, label handling, class-imbalance strategy, per-disaster EDA |
| 2 · Deep Learning Model | 10 | from-scratch **and** pretrained CNN, fair comparison, held-out-disaster eval, reproducibility |
| 3 · Impact & Recovery Analysis | 10 | damage × socioeconomic overlay, report writing |

## Proposed task division (by role, 4 people)

We split by **pipeline stage** so people can work in parallel against a shared `src/` API.
Names are not pre-assigned — pick the role you prefer.

### Role A — Data Engineer  → `01_data_preprocessing.ipynb`
- Download xBD / xView2, parse label JSONs (building polygons + 4-level damage)
- Co-register pre/post tiles, crop a patch per building
- Decide & implement the **class-imbalance** strategy (weights / resampling)
- Produce **per-disaster splits**: train on some disaster types, hold out one entirely
- Deliver a clean `(pre, post, label, disaster_type)` dataset on disk

### Role B — EDA & Impact Analyst  → `02_eda_visualization.ipynb` + `05_impact_recovery.ipynb`
- Class distribution overall and **per disaster type**; example pre/post patches per damage level
- Argue how visually different damage signatures are across disasters (motivates transfer question)
- Part 3: overlay predicted damage with census/socioeconomic data, study unequal impact & recovery
- Lead the written report

### Role C — From-Scratch Model  → `03_model_from_scratch.ipynb`
- Compact **siamese / two-branch** CNN built from scratch
- Co-own the shared training loop & metrics in `src/train_utils.py`

### Role D — Pretrained Model & Evaluation  → `04_model_pretrained_and_eval.ipynb`
- Transfer-learning model on a **pretrained backbone** (e.g. ResNet18) + small head
- Run the **held-out-disaster** evaluation, confusion matrices, scratch-vs-pretrained comparison
- Generate the final **damage map** for a held-out event

### Shared by everyone — `src/`
`config.py` (paths, classes, splits) · `data_utils.py` (Dataset + transforms) ·
`models.py` (both architectures) · `train_utils.py` (train loop + metrics).
Everyone imports from here so the four notebooks stay consistent.
